```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#generar distribución normal
#loc -> media, scale->DE
normal = np.random.normal(loc=50, scale=10, size=1000)
print(normal[:10])

#generar distribución uniforme
#low->limite inferior, high->limite superior
uniforme = np.random.uniform(low=0, high=10, size=1000)
print(uniforme[:10])
```

```python
#calcular media
media_normal = np.mean(normal)
media_uniforme = np.mean(uniforme)
print("Media normal", media_normal)
print("Media uniforme", media_uniforme)

#calcular mediana
mediana_normal = np.median(normal)
mediana_uniforme = np.median(uniforme)
print("Mediana normal", mediana_normal)
print("Mediana uniforme", mediana_uniforme)

#ccalcular varianza
varianza_normal = np.var(normal)
varianza_uniforme = np.var(uniforme)
print("Varianza normal", varianza_normal)
print("Varianza uniforme", varianza_uniforme)

#calcular desvío estándar
desvio_normal = np.std(normal)
desvio_uniforme = np.std(uniforme)
print("Desvio normal", desvio_normal)
print("Desvio uniforme", desvio_uniforme)

#calcular mín y máx
min_normal = np.min(normal)
max_normal = np.max(normal)
min_uniforme = np.min(uniforme)
max_uniforme = np.max(uniforme)
print("Min normal", min_normal)
print("Max normal", max_normal)
print("Min uniforme", min_uniforme)
print("Max uniforme", max_uniforme)
```

```python
#crear gráficos
#plt.subplots->varios gráficos en una fig (cant de filas, cant de colum)
#figsize-> tamaño de la figura
#fig -> figura completa, ax->espacio para el histograma
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

#crear histograma distribución normal
#ax[0] -> primer gráf, bins->divide los datos en intervalos,edgcolor -> color de borde
ax[0].hist(normal, bins=30, color="pink", edgecolor="black", label="Normal")
ax[0].set_title("Distribución Normal")
ax[0].set_xlabel("Valores")
ax[0].set_ylabel("Frecuencia")
ax[0].grid(True, alpha=0.3)
ax[0].legend()

#crear histograma distrib uniforme
ax[1].hist(uniforme, bins=30, color="skyblue", edgecolor="black", label="Uniforme")
ax[1].set_title("Distribución Uniforme")
ax[1].set_xlabel("Valores")
ax[1].set_ylabel("Frecuencia")
ax[1].grid(True, alpha=0.3)
ax[1].legend()

#titulo de la figura
fig.suptitle("Comparación de distribuciones probabílisticas",
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

Reflexión:
Se observan diferencias en la forma y distribución de los datos.
La distribución normal presenta una forma de campana, donde los datos están concentrados al rededor de la media.
La distribución uniforme presenta valores distribuidos de manera relativamente homogenea entre 0 y 10.
El desvio estandar en la distribución normal es 49.47, mientras que en la distribución uniforme es de 2.87, lo que indica una mayor dispersión de valores en la distribución normal.
